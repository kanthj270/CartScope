## Validated definitions

- Exact duplicate records are removed before analysis.
- Missing Amount values use the median of non-cancelled, positive Amount values.
- Zero-amount non-cancelled records remain zero.
- Revenue and units exclude cancelled records.
- AOV is revenue divided by unique non-cancelled orders.
- Returns include both returned-to-seller and returning-to-seller statuses.
